```python
import numpy as np

# Columns: [Age, Math Marks, Science Marks]
data = np.array([
    [18, 85, 78],   # Student 1
    [19, 92, 88],   # Student 2
    [17, 76, 95],   # Student 3
    [18, 65, 70],   # Student 4
    [20, 90, 85]    # Student 5
])
```

```python
#Get the shape of the matrix.
rows = len(s)
columns = len(s[0])

print("Shape of the matrix:", (rows, columns))
print("Shape of the matrix:", data.shape)
```

```python
#Find the average age of students.
average_age = np.mean(data[:, 0])

print("Average age of students:", average_age)
```

```python
#Extract Math marks of all students.
math_marks = data[:, 1]

print("Math marks of all students:", math_marks)
```

```python
#Find the highest Science mark.
science_marks = [student[2] for student in data]

highest_mark = max(science_marks)

print("Highest Science mark:", highest_mark)
```

```python
#Get details of the student who scored more than 90 in Math.
students = data[data[:, 1] > 90]

print("Students scoring more than 90 in Math:")
print(students)
```

```python
#Increase Math marks of all students by 5.
data[:, 1] += 5

print("Updated student data:")
print(data)
```

```python
#Find how many students are younger than 19.
count = np.sum(data[:, 0] < 19)

print("Number of students younger than 19:", count)
```

```python
#Calculate the average marks in each subject (column-wise mean).
average = np.mean(data, axis=0)

print("Average Age:", average[0])
print("Average Math Marks:", average[1])
print("Average Science Marks:", average[2])
```

```python
#Get data of students who scored at least 80 in both subjects.
students = data[(data[:, 1] >= 80) & (data[:, 2] >= 80)]

print("Students who scored at least 80 in both subjects:")
print(students)
```

```python
#Replace all Science marks < 75 with 0.
data[data[:, 2] < 75, 2] = 0

print("Updated student data:")
print(data)
```
